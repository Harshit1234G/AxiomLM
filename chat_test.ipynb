```python
from pathlib import Path
import tensorflow as tf
import sys

axiom_utils = Path('kaggle/input/datasets/harshit1234g/axiomlm-utils')
sys.path.append(str(axiom_utils))
import llm_components as lc
```

```python
sp = lc.load_sp_tokenizer(str(axiom_utils / 'sp_tokenizer.model'))
```

```python
model = tf.keras.models.load_model('kaggle/working/AxiomLM-33M-Instruct.keras')
```

```python
while True:
    prompt = input('Enter your prompt: ')
    if prompt.lower().strip() == 'quit':
        break

    formatted_prompt = f'# Instruction:\n{prompt}\n# Response:\n'
    generated_text = lc.generate_text(
        model,
        tokenizer= sp,
        prompt= formatted_prompt,
        max_new_tokens= 200,
        temperature= 0.8
    )
    response = generated_text.split('# Response:')[1]
    print(f'Response: {response}\n')
```
