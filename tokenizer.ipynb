```python
import sentencepiece as spm
```

```python
train_path = 'kaggle/input/datasets/vadimkurochkin/wikitext-103/wikitext-103/wiki.train.tokens'
out_path = 'kaggle/input/datasets/harshit1234g/axiomlm-utils/sp_tokenizer'
```

```python
spm.set_random_generator_seed(42)
```

```python
spm.SentencePieceTrainer.train(
    input= train_path,
    model_prefix= out_path,
    vocab_size= 16_000,
    model_type= 'bpe',
    unk_id= 0,
    bos_id= 1,
    eos_id= 2,
    pad_id= 3,
    normalization_rule_name= 'identity',
    add_dummy_prefix= False,
    byte_fallback= True,
    num_threads= 2
)
```

```python
sp = spm.SentencePieceProcessor()
sp.load(out_path + '.model')
```

```python
with open(train_path, 'r', encoding= 'utf-8') as f:
    text = f.read()[:1_000_000]
```

```python
tokens = sp.encode(text)
print('Avg chars per token:', len(text) / len(tokens))
```

```python
sp.encode(['This is a test', 'Hello world'], out_type= int)
```

```python
sp.encode(['This is a test', 'Hello world'], out_type= str)
```
